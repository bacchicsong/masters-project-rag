```python
def pdf_to_json_chunks(
    pdf_path: str,
    output_json_path: str,
    chunk_size: int = 1000, 
    chunk_overlap: int = 200,
) -> None:
    reader = PdfReader(pdf_path)
    full_text = ""
    
    for page_num, page in tqdm(enumerate(reader.pages),  total=len(reader.pages)):
        text = page.extract_text()
        if text:
            full_text += text + "\n"

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        length_function=len,
        separators=["\n\n", "\n", " ", ""]
    )

    chunks = text_splitter.split_text(full_text)

    json_data = []
    for index, chunk in tqdm(enumerate(chunks), total=len(chunks)):
        json_data.append({
            "chunk_id": index + 1,
            "text": chunk.strip(),
            "meta": {
                "character_count": len(chunk),
                "source_file": pdf_path
            }
        })

    with open(output_json_path, 'w', encoding='utf-8') as f:
        json.dump(json_data, f, ensure_ascii=False, indent=4)
```

```python
import os
import json
import docx
from tqdm.notebook import tqdm
from pypdf import PdfReader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langdetect import detect, DetectorFactory
from deep_translator import GoogleTranslator

DetectorFactory.seed = 42
DIRECTORY_PATH = "."
OUTPUT_JSON = "parsed_text.json"
CHUNK_SIZE = 1000
CHUNK_OVERLAP = 200

    
# 1. Gather all PDF and DOCX files in the directory
all_files = []
for root, _, files in os.walk(DIRECTORY_PATH):
    for file in files:
        if file.endswith('.pdf') or file.endswith('.docx'):
            all_files.append(os.path.join(root, file))
            
if not all_files:
    print(f"No .pdf or .docx files found in '{DIRECTORY_PATH}'.")

# 2. Initialize the text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=CHUNK_SIZE,
    chunk_overlap=CHUNK_OVERLAP,
    length_function=len,
    separators=["\n\n", "\n", " ", ""]
)

combined_json_data = []
global_chunk_id = 1

# 3. Process each file
for file_path in tqdm(all_files, desc="Processing files"):
    full_text = ""
    
    try:
        if file_path.endswith('.pdf'):
            # Extract text from PDF
            reader = PdfReader(file_path)
            for page in reader.pages:
                text = page.extract_text()
                if text:
                    full_text += text + "\n"
                    
        elif file_path.endswith('.docx'):
            # Extract text from DOCX
            doc = docx.Document(file_path)
            full_text = "\n".join([p.text for p in doc.paragraphs if p.text])
            
    except Exception as e:
        print(f"Skipping {file_path} due to read error: {e}")
        continue

    if not full_text.strip():
        continue

    # 4. Split document text into chunks
    chunks = text_splitter.split_text(full_text)

    # 5. Translate and format each chunk
    for chunk in tqdm(chunks, desc=f"Chunking {os.path.basename(file_path)}", leave=False):
        cleaned_text = chunk.strip()
        if not cleaned_text:
            continue
            
        translated_text = cleaned_text
        is_translated = False
        
        # Detect language and translate if English
        try:
            if detect(cleaned_text) == 'en':
                # Translates English to Russian via Google Translate API
                translated_text = GoogleTranslator(source='en', target='ru').translate(cleaned_text)
                is_translated = True
        except Exception:
            # Fallback to original text if language detection or translation fails
            pass

        combined_json_data.append({
            "chunk_id": global_chunk_id,
            "text": translated_text,
            "meta": {
                "character_count": len(translated_text),
                "original_character_count": len(cleaned_text),
                "source_file": file_path,
                "translated_from_english": is_translated
            }
        })
        global_chunk_id += 1

# 6. Save all gathered chunks into a single JSON file
with open(OUTPUT_JSON, 'w', encoding='utf-8') as f:
    json.dump(combined_json_data, f, ensure_ascii=False, indent=4)

print(f"\nExecution complete! Processed {len(all_files)} files. Saved {len(combined_json_data)} total chunks to {OUTPUT_JSON}")
```

```python
file_path = "tbank_articles.json"
with open(file_path, "rb") as f:
    source_data = json.load(f)
```

```python
combined_json_data = []
global_chunk_id = 0
for doc in tqdm(source_data):
    title = doc.get("title", "").strip()
    
    for section in doc.get("sections", []):
        heading = section.get("heading", "").strip()

        raw_content_list = section.get("content", [])
        if isinstance(raw_content_list, list):
            raw_content = " ".join([str(text).strip() for text in raw_content_list if text])
        else:
            raw_content = str(raw_content_list).strip()

        if not raw_content:
            continue

        cleaned_text = raw_content  
        translated_text = raw_content
        is_translated = False

        context_parts = [title, heading]
        context_prefix = " > ".join([part for part in context_parts if part])
        
        if context_prefix:
            enriched_text = f"{context_prefix}\n\n{translated_text}"
        else:
            enriched_text = translated_text

        global_chunk_id += 1

        combined_json_data.append({
            "chunk_id": global_chunk_id,
            "text": enriched_text,
            "meta": {
                "character_count": len(enriched_text),
                "original_character_count": len(cleaned_text),
                "source_file": file_path,
                "translated_from_english": is_translated
            }
        })

with open("combined_chunks.json", 'w', encoding='utf-8') as f:
    json.dump(combined_json_data, f, ensure_ascii=False, indent=2)
```
